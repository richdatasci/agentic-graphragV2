# Retrieval Debugging


Use this notebook to inspect hybrid retrieval behaviour.

It should help answer:

- Did vector retrieval find semantically useful chunks?
- Did graph retrieval return relevant neighbours?
- Did keyword retrieval catch exact terms?
- Did the reranker improve final context quality?


Retrieval quality should be checked before answer quality. Weak answers are often caused by weak context.


```python
sample_query = "How does faithfulness evaluation work in the Agentic GraphRAG platform?"

vector_results = [
    {"source": "vector", "score": 0.82, "text": "Faithfulness evaluation checks whether claims are supported."}
]

graph_results = [
    {"source": "graph", "score": 0.76, "text": "Faithfulness evaluation EVALUATES Answer."}
]

keyword_results = [
    {"source": "keyword", "score": 0.69, "text": "The faithfulness endpoint returns supported and unsupported claims."}
]

all_results = vector_results + graph_results + keyword_results
all_results
```

```python
sorted_results = sorted(all_results, key=lambda row: row["score"], reverse=True)
sorted_results
```

```python
context = "\n\n".join(
    f"[{idx}] {row['source']} score={row['score']}\n{row['text']}"
    for idx, row in enumerate(sorted_results, start=1)
)

print(context)
```
