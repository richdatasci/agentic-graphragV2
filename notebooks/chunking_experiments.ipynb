# Chunking Experiments


Use this notebook to compare chunking strategies.

Chunking should preserve enough context for faithfulness while keeping retrieval precise.


Experiments to run:

- Fixed-size chunking
- Overlapping chunking
- Section-aware chunking
- Table-aware chunking
- Layout-aware chunking


```python
text = '''
Agentic GraphRAG combines graph retrieval, vector search and multi-agent orchestration.
The ingestion layer extracts entities, relationships, claims and triples.
The evaluation layer checks whether generated answers are supported by retrieved context.
'''.strip()

def chunk_text(value: str, max_chars: int = 80, overlap: int = 20):
    chunks = []
    start = 0
    while start < len(value):
        end = min(len(value), start + max_chars)
        chunks.append(value[start:end])
        if end == len(value):
            break
        start = max(0, end - overlap)
    return chunks

chunk_text(text)
```

```python
for size in [80, 120, 200]:
    chunks = chunk_text(text, max_chars=size, overlap=20)
    print("size", size, "chunks", len(chunks))
```
