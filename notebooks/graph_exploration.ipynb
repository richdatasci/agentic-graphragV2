# Graph Exploration


Use this notebook to inspect extracted graph entities, relationships and triples.

Typical questions:

- Which entities are most connected?
- Which documents contributed most triples?
- Are there duplicate entities?
- Are relationship types useful or too generic?


Expected inputs:

- Neo4j connection details from `.env`
- Extracted triples from the graph database
- Optional exported graph JSON


```python
import os
from collections import Counter

NEO4J_URI = os.getenv("NEO4J_URI", "bolt://localhost:7687")
NEO4J_USER = os.getenv("NEO4J_USER", "neo4j")
NEO4J_PASSWORD = os.getenv("NEO4J_PASSWORD", "agentic-graphrag-password")

print("Neo4j URI:", NEO4J_URI)
```

```python
# Example placeholder triples.
triples = [
    {"subject": "Agentic GraphRAG", "predicate": "combines", "object": "graph retrieval"},
    {"subject": "Agentic GraphRAG", "predicate": "combines", "object": "vector search"},
    {"subject": "Faithfulness evaluation", "predicate": "checks", "object": "answer support"}
]

predicates = Counter(t["predicate"] for t in triples)
predicates
```

```python
import networkx as nx

graph = nx.DiGraph()
for triple in triples:
    graph.add_edge(triple["subject"], triple["object"], label=triple["predicate"])

print("Nodes:", graph.number_of_nodes())
print("Edges:", graph.number_of_edges())
print("Degree:", dict(graph.degree()))
```
